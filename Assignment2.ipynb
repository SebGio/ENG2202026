Grocery Program:

```python
prices = [2, 6, 4, 3, 1]
apples = input("How many apples do you want?: ")
pineapples = input("How many pineapples do you want?: ")
oranges = input("How many oranges do you want?: ")
bananas = input("How many bananas do you want?: ")
strawberries = input("How many strawberries do you want?: ")
total = (int(apples) * prices[0]) + (int(pineapples) * prices[1]) + (int(oranges) * prices[2]) + (int(bananas) * prices[3]) + (int(strawberries) * prices[4])

print("Your total is: $" + str(total))
```

String Program:

```python
sentence = input("Enter a sentence: ")
words = sentence.split()
print("Your words are: " + str(words))
print("The number of words in the sentence is: " + str(len(words)))
```
